# 5. Putting the framework to the test: does the channel carry weight?

**What you get out of this notebook:** the drop-channel ablation of the appendix,
run, and a negative result read for what it is.

The tutorial places each method by the channel its operator is conditioned on. The
rule behind that placement is: remove one channel, hold everything else fixed, and
see whether what the operator emits changes. This notebook runs that removal for
the natural-language channel of an EoH-style operator on bin packing.

```python
import _bootstrap

import json, subprocess, sys

candidates = json.load(open('candidates.json'))
print({k: len(v) for k, v in candidates.items()})
```

`FULL` holds heuristics generated with the parent code, the scores **and** the
natural-language idea (*prefer the bin with the least remaining slack*).
`ABLATED` holds heuristics generated with the parent code and the scores only.
Same evaluator, same instances.

```python
print(subprocess.run([sys.executable, 'ablation.py', 'candidates.json'],
                     capture_output=True, text=True).stdout)
```

## Reading this honestly

Both conditions reach the optimum on every candidate. On this task the
natural-language channel is **not** load-bearing: the operator recovers the
best-fit heuristic with it or without it.

That is a negative result and the tutorial reports it as one. What it licenses is
narrow: on *this* task, with *these* candidates, removing the idea changed nothing
measurable in quality. What it does not license is the general claim that the
channel never matters — the marker line shows the ablated heuristics stopped
expressing the idea, they just did not need it to land on the same rule.

The obvious next experiment is a task where first fit and best fit come apart more
sharply, so that the measurement has room to show a difference. That is the sort of
thing this repository is for.
